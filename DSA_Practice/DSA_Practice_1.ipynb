# (1) Colour sorting problem  (Dutch National Flag problem)

```python
def colour(arr):
    low=0
    mid=0
    high=len(arr)-1
    while mid<=high:
        if arr[mid]==0:
            arr[low],arr[mid]=arr[mid],arr[low]
            low+=1
            mid+=1
        elif arr[mid]==1:
            mid+=1
        else:
            arr[mid],arr[high]=arr[high],arr[mid]
            high-=1
    return arr
ans=colour([2,1,2,1,1,0])
print(ans)
```

## (5) 1ST Method Palindron of string

```python
def palindron(str):
    i=0
    j=len(str)-1
    str=str.lower()
    while i<j:
       if str[i]!=str[j]:
           return f"Not Palindrom"
       i+=1
       j-=1
    return f"Palindrom"
ans=palindron("LAL")
print(ans)
    
    
    
```

## (5) 2nd Method Palindron of string

```python
def palindron_2nd(str):
    temp=str.lower()
    rev=""
    for ch in temp:
        rev=ch+rev
    if temp==rev:
        return f"Palindron"
    else:
        return f"Not Palindron"
ans=palindron_2nd("LAl")
print(ans)
```

# 7(a) Without Repeated longest Substring length

```python
def substring_length(str):
    visited={}
    i=0
    result=0
    for j in range(len(str)):
        if str[j] in visited:
            if i<visited[str[j]]+1:
                i=visited[str[j]]+1
            else:
                i=i
        if result<j-i+1:
            result=j-i+1
        else:
            result=result
        visited[str[j]]=j
    return result
ans=substring_length("anniange")
print(ans)
```

## (9) DNA Problem

```python
def DNA(str):
    str=str.upper()
    lookup={'A':'T','T':'A','C':'G','G':'C'}
    comp=""
    for char in str:
        if char in lookup:
            comp=comp+lookup[char]
        else:
            comp=comp+char
    return comp
ans=DNA("atcg")
print(ans)
```

## (10) Reverse String
  Input: s = ["h","e","l","l","o"]
Output: ["o","l","l","e","h"]

```python
def reverse_string(arr):
    j=len(arr)-1
    i=0
    while i<j:
        arr[i],arr[j]=arr[j],arr[i]
        i+=1
        j-=1
    return arr
ans=reverse_string(["h","e","l","l","o"])
print(ans)
    
```

## (11) Container with most water

```python
def water(arr):
    max_area=float('-inf')
    l=0
    r=len(arr)-1
    while l<r:
        area=min(arr[l],arr[r])*(r-l)
        max_area=max(area,max_area)
        if arr[l]<arr[r]:
            l+=1
        else:
            r-=1
    return max_area
ans=water([1,2,6,10,11,7,4,6])
print(ans)
```

## (13) Bubble sort

```python
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        for j in range(1,n-i):
            if arr[j-1]>arr[j]:
                arr[j-1],arr[j]=arr[j],arr[j-1]
    return arr
ans=bubble_sort([3,9,10,22,100,98,0,-1])
print(ans)
```

## (14) Selection sort

```python
def selection(arr):
    n=len(arr)
    for i in range(n):
        min_index=i
        for j in range(i+1,n):
            if arr[min_index]>arr[j]:
                min_index=j
        arr[min_index],arr[i]=arr[i],arr[min_index]
    return arr
ans=selection([3,9,10,22,100,98,0,-1])
print(ans)
```

## (15) Insertion sort

```python
def insertion_sort(arr):
    n=len(arr)
    for i in range(1,n):
        key=arr[i]
        j=i-1
        while j>=0 and arr[j]>key:
            arr[j+1]=arr[j]
            j-=1
        j+=1
        arr[j]=key
    return arr
ans=insertion_sort([3,9,10,22,100,98,0,-1])
print(ans)
        
```

## (16) Quick sort

```python
def partition(arr,low,high):
    pivot=arr[high]
    i=low-1
    for j in range(low,high):
        if arr[j]<=pivot:
            i+=1
            arr[i],arr[j]=arr[j],arr[i]
    arr[i+1],arr[high]=arr[high],arr[i+1]
    return i+1
def quick_sort(arr,low,high):
    if low<high:
        pi=partition(arr,low,high)
        quick_sort(arr,low,pi-1)
        quick_sort(arr,pi+1,high)
        
    return arr
arr=[3,9,10,22,100,98,0,-1]
ans=quick_sort(arr,0,len(arr)-1)
print(ans)
            
```

## (17) Merge sort

```python
def merge_sort(arr):
    if len(arr)<=1:
        return arr
    mid=len(arr)//2
    left=merge_sort(arr[:mid])
    right=merge_sort(arr[mid:])
    return merge(left,right)
def merge(left,right):
    i=0
    j=0
    result=[]
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1
    while i<len(left):
        result.append(left[i])
        i+=1
    while j<len(right):
        result.append(right[j])
        j+=1
    return result
ans=merge_sort([3,9,10,22,100,98,0,-1])
print(ans)
    
```

## (21) Sum of Unique Elements

```python
def unique_sum(arr):
    n=len(arr)
    visited={}
    for i in range(n):
        if arr[i] not in visited:
            visited[arr[i]]=1
        else:
            visited[arr[i]]+=1
    sum=0
    for key,value in visited.items():
        if value==1:
            sum+=key
    return sum
ans=unique_sum([1,2,3,4,5,1,2,3,4])
print(ans)
        
```

## (24) No of subarrayArray with Target sum 

```python
def TargetSubArray(arr,k):
    visited={0:1}
    curr_sum=0
    count=0
    for num in arr:
        curr_sum+=num
        diff=curr_sum-k
        if diff in visited:
            count+=visited[diff]
        if curr_sum in visited:
            visited[curr_sum]+=1
        else:
            visited[curr_sum]=1
    return count
ans=TargetSubArray([1,2,4,7,-1,-2,3,4],7)
print(ans)
```

## (25) All subarrayArray with Target sum 

```python
def AllSubArr(arr,k):
    prefix_map={0:[-1]}
    curr_sum=0
    result=[]
    count=0
    for i ,nums in enumerate(arr):
        curr_sum+=nums
        diff=curr_sum-k
        if diff in prefix_map:
            for start in prefix_map[diff]:                
                result.append(arr[start+1:i+1])
                count+=1
        if curr_sum in prefix_map:
            prefix_map[curr_sum].append(i)
        else:
            prefix_map[curr_sum]=[i]
    return count,result
ans=AllSubArr([1,2,4,7,-1,-2,3,4],7)
print(ans)
            
```

## (29) Two Sum Index

```python
def TwoSum(arr,Target):
    count={}
    for i,num in enumerate(arr):
        if Target-num in count:
            return [count[Target-num],i]
        count[num]=i
ans=TwoSum([1,2,5,9,8,6],7)
print(ans)
```

## (32) Valid Anagram

```python
def Valid_Anagram(s1,s2):
    s1=s1.upper()
    s2=s2.upper()
    count={}
    if len(s1)!=len(s2):
        return False
    for char in s1:
        if char in count:
            count[char]+=1
        else:
            count[char]=1
    for char in s2:
        if char not in count or count[char]==0:
            return False
        count[char]-=1
    return True
ans=Valid_Anagram("ANAa","NaAAi")
print(ans)
```

## (33) Palindrome Number

```python
def palindrome(x):
    original=x
    if x<0 or (x!=0 and x%10==0):
        return False
    rev=0
    while x>0:
        rev=rev*10+x%10
        x//=10
    if rev==original:
        return True
ans=palindrome(1110)
print(ans)        
```

## (34) strs = ["flower","flow","flight"]


```python
def longest_common_prefix(arr):
    ans=""
    v=sorted(arr)
    first=v[0]
    last=v[-1]
    for i in range(min(len(first),len(last))):
        if first[i]!=last[i]:
            return ans
        ans+=last[i]
    return ans
answer=longest_common_prefix(["flower","flow","flight"])
print(answer)        
    
```

## (35) Add Digits

```python
def addDigit(num):
    if num<0:
        num=(-1)*num
    while num>10:
        sum_digit=0
        while num>0:            
            sum_digit+=num%10
            num//=10
        num=sum_digit
    return num
ans=addDigit(-12)
print(ans)
```

## (38) Write a program that will take three digits from the user and add the square of each digit.

```python
def digit_square(num):
    if num>0 and num<10:
        return num*num
    else:
        sum=0
        while num>0:
            digit=num%10
            sum+=digit**2
            num//=10
        return sum
ans=digit_square(26)
print(ans)
```

## (39) Write a program that will check whether the number is armstrong number or not.


```python
def ArmstrongNumber(t):
    if t<0:
        return False
    num=str(t)
    n=len(num)
    sum=0
    for i in num:
        sum+=int(i)**n
    if sum==t:
        return True
    else:
        return False
ans=ArmstrongNumber(153)
print(ans)
```

## (40) Write a program that will Convert Factorial


```python
def fact(n):
    if n<0:
        return f"Number is Negative so factorial is not defined"
    elif n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
ans=fact(-3)
print(ans)
```
