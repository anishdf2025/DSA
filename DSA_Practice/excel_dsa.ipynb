# (30) Given an array of size n and a number k, fin all elements that appear more than " n/k " times.

```python
def count_n_k(arr,k):
    n=len(arr)
    x=n//k
    freq={}
    for num in arr:
        if num in freq:
            freq[num]+=1
        else:
            freq[num]=1
    count=0
    for key,value in freq.items():
        if value>x:
            print("key",key,"----->count",value)
            count+=1
    return count

arr=[1,3,4,4,4,6,7,2,2,6,7,6,7,7]
count=count_n_k(arr,5)
print(count)            
```

## (31) Maximum profit by buying and selling a share atmost twice

```python
def max_profit(price):
    n=len(price)
    if n==0:
        return 0
    left=[0]*n
    min_price=price[0]
    for i in range(1,n):
        min_price=min(min_price,price[i])
        profit=price[i]-min_price
        left[i]=max(profit,left[i-1])
    right=[0]*n
    max_price=price[-1]
    for i in range(n-2,-1,-1):
        max_price=max(max_price,price[i])
        profit=max_price-price[i]
        right[i] =max(profit,right[i+1])
    ans=0
    for i in range(n):
        ans=max(ans,left[i]+right[i])
    return ans
price=[10,1,20,40,100]
ans=max_profit(price)
print(ans)
```

 # 37 Three way partitioning of an array around a given value

```python
def three_partition(arr,a,b):
    low=0
    mid=0
    high=len(arr)-1
    while mid<=high:
        if arr[mid]<a:
            arr[low],arr[mid]=arr[mid],arr[low]
            low+=1
            mid+=1
        elif arr[mid]>b:
            arr[mid],arr[high]=arr[high],arr[mid]
            high-=1
        else:
            mid+=1
    return arr
arr=[1,5,9,10,4,7,7]
ans=three_partition(arr,5,7)
print(ans)
```

# 38 Minimum swaps required bring elements less equal K together

```python
def min_swaps(arr,k):
    n=len(arr)
    good=0
    for i in range(n):
        if arr[i]<=k:
            good+=1
    if good<1 or n<1:
        return 0
    ans=float('inf')
    for start in range(n-good+1):
        bad=0
        for i in range(start,start+good):
            if arr[i]>k:
                bad+=1
        ans=min(ans,bad)
    return ans
arr = [2, 7, 9, 5, 8, 7, 4]
k = 6

print(min_swaps(arr, k))
            
        
```

# (107) Maximum and minimum of an array using minimum number of comparisons

```python
def middle_1(a,b,c):
    arr=[a,b,c]
    arr.sort()
    return arr[1]
a,b,c=1,4,3 
ans=middle_1(a,b,c)
print(ans)   
```

```python
def middle_2(a,b,c):
    if (a<b<c) or (c<b<a):
        return b
    elif (b<a<c) or (c<a<b):
        return a
    else:
        return c
a,b,c=1,6,2
ans=middle_2(a,b,c)
print(ans)
```

```python
def middle_3(a,b,c):
    return (a+b+c)-max(a,b,c)-min(a,b,c)
a,b,c=9,10,6
ans=middle_3(a,b,c)
print(ans)
```

# (109) Find the repeating and the missing

```python
def miss_rep(arr):
    n=len(arr)
    for i in range(n):
        idx=abs(arr[i])-1
        if arr[idx]<0:
            duplicate=abs(arr[i])
        else:
            arr[i]*=-1
    for i in range(n):
        if arr[i]>0:
            miss=i+1
    return [duplicate,miss]
arr=[1,1,3]
dup,miss=miss_rep(arr)
print(dup,miss)
        
```

# (111) Searching in an array where adjacent differ by at most k

```python
def adjacent_index(arr,k,x):
    i=0
    while i<(len(arr)):
        if arr[i]==x:
            return i
        diff=abs(arr[i]-x)
        jump=diff//k
        index=max(1,jump)
        i=i+index
    return -1
arr=[1,2,3,4,5,4,4,6]
ans=adjacent_index(arr,1,6)
print(ans)
```
