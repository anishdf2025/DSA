## 43 Matrix Transpose

```python
def Matrix_Transpose(mat):
    n=len(mat)
    for i in range(n):
        for j in range(i+1,n):
            mat[i][j],mat[j][i]=mat[j][i],mat[i][j]
    return mat
mat=[[1,2,3],
     [4,5,6],
     [7,8,9]
     ]
ans=Matrix_Transpose(mat)
print(ans)
    
```

## (44) First Index([0][0]) SpiralOrder clockwise

```python
def spiralOrder(matrix):
        row,column=len(matrix),len(matrix[0])
        result=[]
        x,y,dx,dy=0,0,1,0
        for _ in range(row*column):
            result.append(matrix[y][x])
            matrix[y][x]="."

            if not 0<=x+dx<column or not 0<=y+dy<row or matrix[y+dy][x+dx]==".":
                dx,dy=-dy,dx
            x+=dx
            y+=dy
        return result
    
mat=[[1,2,3],
     [4,5,6],
     [7,8,9]
     ]
ans=spiralOrder(mat)
print(ans)
```

## (46) Diagonal Sum

```python
def DiagonalSum(matrix):
    sum=0
    i,j=0,0
    while i<len(matrix) and j<len(matrix[0]):
        if i==j:
            sum+=matrix[i][j]
        i+=1
        j+=1
    return sum
mat=[[1,2,3],
     [4,5,6],
     [7,8,0]
     ]
ans=DiagonalSum(mat)
print(ans)
```

## (48) Reverse a List

```python
def Reverse_List(list):
    left,right=0,len(list)-1
    while left<right:
        list[left],list[right]=list[right],list[left]
        left+=1
        right-=1
    return list
list=[1,2,3,4,5,6,7,8,9,0]
ans=Reverse_List(list)
print(ans)
        
    
```

## (50) Binary Search

```python
def Binary_Search(arr,target):
    left=0
    right=len(arr)-1
    arr.sort()
    while left<right:
        mid=(left+right)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]>target:
            right=mid-1
        else:
            left=mid+1
    return False
number = [10, 20, 30, 40, 50]
target = 31
Binary_Search(number,target)
```

## (51) Write a python program to remove all the duplicates from a list

```python
def Remove_Duplicates_list(arr):
    arr.sort()
    seen=set()
    result=[]
    for num in arr:
        if num not in seen:
            seen.add(num)
            result.append(num)
    return result
number = [10, 20, 30, 40, 50,1,1,10,20,2,2,2,2]
ans=Remove_Duplicates_list(number)
print(ans)

    
```

## (52) only Duplicate Elements

```python
def Duplicate(arr):
    seen=set()
    duplicate=set()
    for num in arr:
        if num not in arr:
            seen.add(num)
        else:
            duplicate.add(num)
    return duplicate
number = [10, 20, 30, 40, 50,1,1,10,20,2,2,2,2]
ans=Duplicate(number)
print(ans)
```

## (55) Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

```python
def calculate_average():
    total_sum = 0
    count = 0

    while True:
        user_input = input("Enter a number (type 'exit' to stop): ")

        if user_input.lower() == "exit":
            break

        number = int(user_input)   # convert AFTER checking exit
        total_sum += number
        count += 1

    if count == 0:
        return 0

    average = total_sum / count
    return average


result = calculate_average()
print("Average:", result)
```

# 87 Write a function that accepts a number and returns it’s factorial. You can not use any loop

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
ans=factorial(5)
print(ans)
```

# 66 Write a program to check if a list is in ascending order or not

```python
def is_ascending(arr):
    for i in range(len(arr)-1):
        if arr[i]>arr[i+1]:
            return False
    return True
ans =is_ascending([1,2,1,4,5,6])
print(ans)
```
