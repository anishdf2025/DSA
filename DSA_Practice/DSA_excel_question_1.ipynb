# Write a Program to check whether a string is a valid shuffle of two strings or not

```python
# 🔹 Input
s1 = input("Enter s1: ")
s2 = input("Enter s2: ")
s3 = input("Enter s3: ")

# 🔹 Length check
if len(s1) + len(s2) != len(s3):
    print(False)
else:
    def solve(i, j, k):
        # Base case
        if k == len(s3):
            if i == len(s1) and j == len(s2):
                return True
            else:
                return False

        # Take from s1
        if i < len(s1) and s1[i] == s3[k]:
            if solve(i + 1, j, k + 1):
                return True

        # Take from s2
        if j < len(s2) and s2[j] == s3[k]:
            if solve(i, j + 1, k + 1):
                return True

        return False

    result = solve(0, 0, 0)
    print(result)
```
