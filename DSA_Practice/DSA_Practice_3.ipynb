## Write a program that will give you the sum of 3 digits

```python
def digit_square(num):
    if num>0 and num<10:
        return num
    else:
        sum=0
        while num>0:
            digit=num%10
            sum+=digit
            num//=10
        return sum
ans=digit_square(267)
print(ans)
```

## Write a program that will reverse a four digit number.Also it checks whether the reverse is true.

```python
def digit_rev(num):
    if num>0 and num<10:
        return num
    sum=0
    origin=num
    while num>0:
        digit=num%10
        sum=sum*10+digit
        num//=10
    if sum==origin:
        return f"palindrome"
    else:
        return f"Number is not palindrom"
ans=digit_rev(121)
print(ans)
            
```

## Example of Join Syntex

```python
list=["ANISH","KUMAR","MAURYA","SON","OF","RAM PRAKASH MAURYA"]
Phrase=" ".join(list)
print(Phrase)
```

```python
def overlap_chunking(text,chunk_size,overlap_size):
    word=text.split()
    result=[]
    step=chunk_size-overlap_size
    for i in range(0,len(word),step):
        chunk_p=word[i:i+chunk_size]
        join_phrase=" ".join(chunk_p)
        result.append(join_phrase)
        if i+chunk_size>=len(word):
            break
    return result
text="ANISH KUMAR MAURYA SON OF RAM PRAKASH MAURYA AND BROTHER AND SISTER NAME ANULATA MANURYA NITISH KUMAR MAURYA"
ans=overlap_chunking(text,6,2)
print(ans)
```
