# Step 1 — Basic Dynamic Prompt

```python
def create_prompt(name, topic):
    prompt = f"""
You are an interviewer.

Candidate name: {name} 
Topic: {topic}

Ask one interview question about {topic}.
Do not provide the answer.
"""
    return prompt


result = create_prompt("Anish", "Python")

print(result)
```

## Step 2 — Add Dynamic Difficulty

```python
def create_propmpt(name,topic,difficulty):
    prompt=f"""
    You are Interviewer. 
    Candidate name :{name} 
    Topic name :{topic} 
    Difficulty :{difficulty} 
     
    Ask the {difficulty} question from this {topic} 
    Do Not give the answer 
    """     
    return prompt 
ans=create_propmpt("Anish","Python","simple")
print(ans)
```
