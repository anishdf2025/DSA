# (1) Colour sorting problem  (Dutch National Flag problem)

```python
def colour(arr):
    low=0
    mid=0
    high=len(arr)-1
    while mid<=high:
        if arr[mid]==0:
            arr[low],arr[mid]=arr[mid],arr[low]
            low+=1
            mid+=1
        elif arr[mid]==1:
            mid+=1
        else:
            arr[mid],arr[high]=arr[high],arr[mid]
            high-=1
    return arr
ans=colour([2,1,2,1,1,0])
print(ans)
```

## (5) 1ST Method Palindron of string

```python
def palindron(str):
    i=0
    j=len(str)-1
    str=str.lower()
    while i<j:
       if str[i]!=str[j]:
           return f"Not Palindrom"
       i+=1
       j-=1
    return f"Palindrom"
ans=palindron("LAL")
print(ans)
    
    
    
```

## (5) 2nd Method Palindron of string

```python
def palindron_2nd(str):
    temp=str.lower()
    rev=""
    for ch in temp:
        rev=ch+rev
    if temp==rev:
        return f"Palindron"
    else:
        return f"Not Palindron"
ans=palindron_2nd("LAl")
print(ans)
```

## (6(a)) Count Frequency character from word 

```python
def frequency_counter(str):
    count={}
    for ch in str:
        if ch in count:
            count[ch]+=1
        else:
            count[ch]=1
    return count
ans=frequency_counter("anniange")
print(ans)
```

# 7(a) Without Repeated longest Substring length

```python
def substring_length(str):
    visited={}
    i=0
    result=0
    for j in range(len(str)):
        if str[j] in visited:
            if i<visited[str[j]]+1:
                i=visited[str[j]]+1
            else:
                i=i
        if result<j-i+1:
            result=j-i+1
        else:
            result=result
        visited[str[j]]=j
    return result
ans=substring_length("anniange")
print(ans)
```

# 7(b) Without Repeated longest Substring 

```python
def substring(str):
    visited={}
    result=0
    i=0
    start=0
    for j in range(len(str)):
        if str[j] in visited:
                if i<visited[str[j]]+1:
                    i=visited[str[j]]+1
                else:
                    i=i
        if result<j-i+1:
            result=j-i+1
            start=i
        else:
            result=result
        visited[str[j]]=j
    return f"{str[start:start+result]}"
ans=substring("anishanigeanidjdjuwtw")
print(ans)
    
```

## (8) Count INCLUDING overlapping

```python
def count_substring(s1,s2):
    count=0
    for i in range(len(s1)-len(s2)+1):
        if s1[i:i+len(s2)]==s2:
            count+=1
    return count
ans=count_substring("anianajasjkdjcksdcjksdbkcbhsldchjanihjdcbbdbcdfbucacbujdfaanidakjcfdbjani","ani")
print(ans)
```

## (9) DNA Problem

```python
def DNA(str):
    str=str.upper()
    lookup={'A':'T','T':'A','C':'G','G':'C'}
    comp=""
    for char in str:
        if char in lookup:
            comp=comp+lookup[char]
        else:
            comp=comp+char
    return comp
ans=DNA("atcg")
print(ans)
```

## (10) Reverse String
  Input: s = ["h","e","l","l","o"]
Output: ["o","l","l","e","h"]

```python
def reverse_string(arr):
    j=len(arr)-1
    i=0
    while i<j:
        arr[i],arr[j]=arr[j],arr[i]
        i+=1
        j-=1
    return arr
ans=reverse_string(["h","e","l","l","o"])
print(ans)
    
```
