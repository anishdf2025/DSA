```python
d = {'A': 100, 'c': 20, 'b': 30}
result=sum(d.values())
print(result)
```

```python
result=sum(d.values())/len(d)
print(result)
```

```python
sort=dict(sorted(d.items()))
print(sort)
```

```python
sort=dict(sorted(d.items(),key=lambda x:x[1]))
print(sort)
```

```python
max_value=max(d,key=d.get)
print(max_value)
```

```python
min_value=min(d,key=d.get)
print(min_value)
```

```python
top_2_max=sorted(d.items(),key=lambda x:x[1],reverse=True)[:2]
print(top_2_max)
```

# Average of list in key of dictionary

```python
d2 = {'A':[10,20,30], 'B':[5,5,6], 'C':[40,11]}
```

```python
result={}
for k,v in d2.items():
    total=sum(v)
    avg=total/len(v)
    result[k]=avg
print(result)
```

```python
result={}
for k,v in d2.items():
    total=max(v)
    result[k]=total
print(result)
```

```python
result={}
for k,v in d2.items():
    total=min(v)
    result[k]=total
    min_pair=sorted(result.items(),key=lambda x:x[1])[-1]
print(min_pair)
```

```python
d3 = {
    'A': [10, 20, 30],
    'B': [40, 50],
    'C': [15, 25, 35],
    'D': [5, 10]
}
```
