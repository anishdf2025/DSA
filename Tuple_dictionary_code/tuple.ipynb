```python
data = [
    ('A', [10, 20, 30]),
    ('B', [5, 5, 5]),
    ('C', [40, 10]),
    ('D', [1, 2, 3, 4])
]
```

```python
result=[]
for key,value in data:
    avg=sum(value)/len(value)
    result.append((key,avg))
print(sorted(result,key=lambda x:x[1],reverse=True)[-1])
```
