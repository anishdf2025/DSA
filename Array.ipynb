# reverse array

```python
def rev(array):
    n=len(array)
    tem=[0]*n
    for i in range(n):
        tem[n-i-1]=array[i]
    return tem
array=[1,2,3,4,5,-2,-3]
result=rev(array)
print(result)
    
```

```python
def revs(array):
    j=len(array)-1
    i=0
    while i<j:
        array[i],array[j]=array[j],array[i]
        i+=1
        j-=1
    return array        
array=[1,2,3,4,5,-2,-3]
result=revs(array)
print(result)
```

Max and min element

```python
def min_max(array):
    min=float('inf')
    max=float('-inf')
    for num in array:
        if num<min:
            min=num
        if num>max:
            max=num
    return min ,max
array=[-1,-1.1,1,10,1000]
min,max=min_max(array)
print(min,max)
```

Sort array with three unique elements(Dutch flag Algorithms)

```python
array=[2,1,0,1,2,0,0,0,1]

def dutch_012(arr):
    low=0
    mid=0
    high=len(array)-1
    while mid<=high:
        if arr[mid]==0:
            arr[low],arr[mid]=arr[mid],arr[low]
            low+=1
            mid+=1
        elif arr[mid]==1:
            mid+=1
        else:
            arr[mid],arr[high]=arr[high],arr[mid]
            high-=1
    return arr
result=dutch_012(array)

print(result)
```

```python
def min_jumps(arr):
    n = len(arr)

    # If array has 0 or 1 element → no jumps needed
    if n <= 1:
        return 0

    # If first element is 0 → can't move anywhere
    if arr[0] == 0:
        return -1

    jumps = 0        # total jumps taken
    curr_end = 0     # current reachable range end
    farthest = 0     # farthest index reachable from current range

    for i in range(n - 1):
        farthest = max(farthest, i + arr[i])

        if i == curr_end:     # time to take a jump
            jumps += 1
            curr_end = farthest

            if curr_end >= n - 1:   # already reached last index
                return jumps

    return -1
```
