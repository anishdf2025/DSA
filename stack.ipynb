```python
def valPar(str):
    stack=[]
    par={')':'(',']':'[','}':'{'}
    
    for ch in str:
        if ch in "{[(":
            stack.append(ch)
        elif ch in ")}]":
            if not stack or stack.pop()!=par[ch]:
                return False
    return len(stack)==0
print(valPar("{[()]})"))   # True

    
```

```python
def sub(s):
    n=len(s)
    for mask in range(1<<n):
        curr=""
        for i in range(n):
            if mask & (1<<i):
                curr+=s[i]
        print(curr)
sub("abcd")
                
        
```
