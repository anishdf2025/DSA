```python
mat=[[1,2,3],[4,5,6],[7,8,9]]
n=len(mat)
def matrix_r_90(matrix):
    for i in range(n):
        for j in range(i+1,n):
            mat[i][j],mat[j][i]=mat[j][i],mat[i][j]
            
    for row in matrix:
        row.reverse()
    return matrix
matrix_r_90(mat)
matrix_r_90(mat)
matrix_r_90(mat)

print(mat)
```

```python
mat = [
    [1, 2, 3, 4],
    [4, 5, 6, 8],
    [7, 8, 9, 10]
]

m=len(mat)
n=len(mat[0])
result=[[0]*m for _ in range(n)]
def rot_90(mat):
    for i in range(m):
        for j in range(n):
            result[j][m-i-1]=mat[i][j]
    return result
results=rot_90(mat)
print(results)
```

# (44) Spiral Print Matrix

```python
def spiral(matrix):
    result=[]
    top=0
    bottom=len(matrix)-1
    left=0
    right=len(matrix[0])-1
    while left<=right and top<=bottom:
        for i in range(left,right+1):
            result.append(matrix[top][i])
        top+=1
        for i in range(top,bottom+1):
            result.append(matrix[i][right])
        right-=1
        for i in range(right,left-1,-1):
            result.append(matrix[bottom][i])
        bottom-=1
        for i in range(bottom,top-1,-1):
            result.append(matrix[i][left])
        left+=1
        
    return result

matrix=[
    [1,2,3,4],
    [4,5,6,8],
    [7,8,9,10],
    [1,2,3,4]
]
ans=spiral(matrix)
print(ans)
```

```python
def spiral(matrix):
    result = []
    top = 0
    bottom = len(matrix) - 1
    left = 0
    right = len(matrix[0]) - 1

    while left <= right and top <= bottom:

        for i in range(left, right + 1):
            result.append(matrix[top][i])
        top += 1

        for i in range(top, bottom + 1):
            result.append(matrix[i][right])
        right -= 1

        # ✅ ADD THIS CONDITION
        if top <= bottom:
            for i in range(right, left - 1, -1):
                result.append(matrix[bottom][i])
            bottom -= 1

        # ✅ ADD THIS CONDITION
        if left <= right:
            for i in range(bottom, top - 1, -1):
                result.append(matrix[i][left])
            left += 1

    return result


matrix = [
    [1, 2, 3, 4],
    [4, 5, 6, 8],
    [7, 8, 9, 10]
]

ans = spiral(matrix)
print(ans)
```

```python
def repeated_number(mat):
    count={}
    for i in range(len(mat[0])):
        count[mat[0][i]]=1
    for i in range(1,len(mat)):
         seen=set()
         for j in range(len(mat[0])):
             if mat[i][j] in count and mat[i][j] not in seen:
                 count[mat[i][j]]+=1
                 seen.add(mat[i][j])
    result=[]
    for element in count:
        if count[element]==len(mat):
            result.append(element)
    print(result)
mat=[[1,2,4,5],[2,11,1,33]]
repeated_number(mat)
```
